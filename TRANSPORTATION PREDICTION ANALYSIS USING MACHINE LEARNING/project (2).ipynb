```python
### encoding
data=pd.get_dummies(df.Vehicle_Type,prefix='Vehicle_Type',drop_first=True)
df=pd.concat([data,df],axis=1)
df.drop(columns='Vehicle_Type',inplace=True)
```

```python
data=pd.get_dummies(df.Traffic_Density,prefix='Traffic_Density',drop_first=True)
df=pd.concat([data,df],axis=1)
df.drop(columns='Traffic_Density',inplace=True)
```

```python
data=pd.get_dummies(df.Weather,prefix='Weather',drop_first=True)
df=pd.concat([data,df],axis=1)
df.drop(columns='Weather',inplace=True)
```

```python
data=pd.get_dummies(df.Road_Condition,prefix='Road_Condition',drop_first=True)
df=pd.concat([data,df],axis=1)
df.drop(columns='Road_Condition',inplace=True)
```

```python
data=pd.get_dummies(df.Accident_Occurred,prefix='Accident_Occurred',drop_first=True)
df=pd.concat([data,df],axis=1)
df.drop(columns='Accident_Occurred',inplace=True)
```

```python
data=pd.get_dummies(df.Accident_Severity,prefix='Accident_Severity',drop_first=True)
df=pd.concat([data,df],axis=1)
df.drop(columns='Accident_Severity',inplace=True)
```

```python
data=pd.get_dummies(df.Alert_Generated,prefix='Alert_Generated',drop_first=True)
df=pd.concat([data,df],axis=1)
df.drop(columns='Alert_Generated',inplace=True)
```

```python
### feature scaling
from sklearn.preprocessing import MinMaxScaler
scale=MinMaxScaler()
df[['Latitude','Longitude','Vehicle_Count','Avg_Speed(km/h)','Visibility(m)','temperature','humidity']]=scale.fit_transform(df[['Latitude','Longitude','Vehicle_Count','Avg_Speed(km/h)','Visibility(m)','temperature','humidity']])
```

```python
##defining x values
x=df.iloc[:,:]
```

```python
### train-test splitting of data
import pandas as pd
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=12)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
### model creation
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
## model evaluation
from sklearn.metrics import accuracy_score,precision_score,classification_report
y_pred=model.predict(x_test)
print(accuracy_score(y_pred,y_test))
print(precision_score(y_pred,y_test))
print(classification_report(y_pred,y_test))
```
